```python
from transformers import pipeline

# Initialize the pipeline
extractor = pipeline("feature-extraction", model="bert-base-uncased")

# Get the vector for a specific word
word = "king"
result = extractor(word)

# The output is a list containing the vector
# Shape: [1, sequence_length, hidden_size]

#print(f"Vector for '{word}': {result[0][1][:5]}...") # Showing first 5 dimensions
v = result[0][1]
print(v)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from transformers import pipeline
import numpy as np

extractor = pipeline("feature-extraction", model="bert-base-uncased")

word_1 = "Australia"
word_2 = "Australian"
v_1 = np.array(extractor(word_1)[0][1]) # Convert to numpy array
v_2 = np.array(extractor(word_2)[0][1]) # Convert to numpy array

word_3 = "Cambodia"
word_4 = "Cambodian"
v_3 = np.array(extractor(word_3)[0][1]) # Convert to numpy array
v_4 = np.array(extractor(word_4)[0][1]) # Convert to numpy array

word_5 = "Australia"
word_6 = "Aussie"
v_5 = np.array(extractor(word_5)[0][1]) # Convert to numpy array
v_6 = np.array(extractor(word_6)[0][1]) # Convert to numpy array

print(cosine_similarity([v_1 - v_2], [v_3 - v_4])[0][0])
print(cosine_similarity([v_5 - v_6], [v_3 - v_4])[0][0])






# result = extractor(word)
# v = np.array(result[0][1]) # Convert v to numpy array

# # words = ["wine","beer","surgar", "corn", "petroleum", "muslim", "colonization", "tariff"]
# # words = ["wine"]

# # words = ["microwave", "refridgerator", "tv", "mother", "often", "frequently", "rarely"]
# words = ["often", "frequently", "rarely", "randomly"]

# sim_dict ={}
# for word in words:
# #   u = np.array(extractor(word)[0][1]) # Convert u to numpy array
#   print(word)
#   print(cosine_similarity([v], [u])[0][0])
```
